```python
def clean_location_columns(df):
    # State cleaning
    df['state_clean'] = (
        df['state']
        .astype(str)
        .str.strip()
        .str.title()
    )

    # District cleaning
    df['district_clean'] = (
        df['district']
        .astype(str)
        .str.strip()
        .str.title()
    )

    # Mark numeric contamination
    df['state_is_numeric'] = df['state_clean'].str.isnumeric()
    df['district_is_numeric'] = df['district_clean'].str.isnumeric()

    # Replace numeric-only entries with NaN in clean columns
    df.loc[df['state_is_numeric'], 'state_clean'] = pd.NA
    df.loc[df['district_is_numeric'], 'district_clean'] = pd.NA

    return df
```

```python
enrol_df = clean_location_columns(enrol_df)
demo_df  = clean_location_columns(demo_df)
bio_df   = clean_location_columns(bio_df)
```

```python
def mark_valid_location(df):
    df['valid_location'] = (
        df['state_clean'].notna() &
        df['district_clean'].notna()
    )
    return df
```

```python
enrol_df = mark_valid_location(enrol_df)
demo_df  = mark_valid_location(demo_df)
bio_df   = mark_valid_location(bio_df)
```

```python
print("Invalid location rows:")
print("Enrolment:", (~enrol_df['valid_location']).sum())
print("Demographic:", (~demo_df['valid_location']).sum())
print("Biometric:", (~bio_df['valid_location']).sum())
```

```python
import re
def normalize_state_name(state):
    """
    Standardizes Indian state/UT names by:
    - Lowercasing
    - Replacing '&' with 'and'
    - Removing extra spaces
    """
    if pd.isna(state):
        return pd.NA

    state = state.lower().strip()
    state = state.replace("&", "and")
    state = " ".join(state.split())

    return state
```

```python
for df in [enrol_df, demo_df, biometric_df]:
    df['state_norm'] = df['state'].apply(normalize_state_name)
```

```python
STATE_CANONICAL_MAP = {
    # Andaman & Nicobar
    "andaman and nicobar islands": "Andaman and Nicobar Islands",

    # Dadra & Nagar Haveli / Daman & Diu
    "dadra and nagar haveli": "Dadra and Nagar Haveli and Daman and Diu",
    "daman and diu": "Dadra and Nagar Haveli and Daman and Diu",
    "dadra and nagar haveli and daman and diu": "Dadra and Nagar Haveli and Daman and Diu",

    # Jammu & Kashmir / Ladakh
    "jammu and kashmir": "Jammu and Kashmir",

    # Delhi
    "delhi": "Delhi",

    # Puducherry
    "pondicherry": "Puducherry",

    # Odisha
    "orissa": "Odisha"
}
```

```python
def apply_canonical_mapping(state):
    if pd.isna(state):
        return pd.NA
    return STATE_CANONICAL_MAP.get(state, state.title())
```

```python
for df in [enrol_df, demo_df, biometric_df]:
    df['state_clean'] = df['state_norm'].apply(apply_canonical_mapping)
```

```python
print("Unique states after cleaning:")
print("Enrolment:", enrol_df['state_clean'].nunique())
print("Demographic:", demo_df['state_clean'].nunique())
print("Biometric:", biometric_df['state_clean'].nunique())
```

```python
def find_suspicious_states(df, name):
    print(f"\nSuspicious states in {name}:")
    suspicious = (
        df['state_clean']
        .dropna()
        .loc[df['state_clean'].str.contains(r'\d', regex=True)]
        .unique()
    )
    print(suspicious[:20], " ... total:", len(suspicious))


find_suspicious_states(enrol_df, "Enrolment")
find_suspicious_states(demo_df, "Demographic")
find_suspicious_states(biometric_df, "Biometric")
```

```python
def is_valid_state_name(state):
    if pd.isna(state):
        return False
    if any(char.isdigit() for char in state):
        return False
    return True
```

```python
for df in [enrol_df, demo_df, biometric_df]:
    df['valid_state'] = df['state_clean'].apply(is_valid_state_name)
```

```python
print("Valid unique states only:")
print("Enrolment:", enrol_df.loc[enrol_df['valid_state'], 'state_clean'].nunique())
print("Demographic:", demo_df.loc[demo_df['valid_state'], 'state_clean'].nunique())
print("Biometric:", biometric_df.loc[biometric_df['valid_state'], 'state_clean'].nunique())
```

```python
for df in [enrol_df, demo_df, biometric_df]:
    df['analysis_ready'] = (
        df['has_valid_date'] &
        df['valid_state'] &
        df['valid_location']  # district + state both valid
    )
```

```python
def normalize_district_name(district):
    if pd.isna(district):
        return pd.NA

    district = district.lower().strip()
    district = district.replace("&", "and")
    district = " ".join(district.split())

    return district
```

```python
for df in [enrol_df, demo_df, biometric_df]:
    df['district_norm'] = df['district'].apply(normalize_district_name)
```

```python
def is_valid_district_name(district):
    if pd.isna(district):
        return False
    if any(char.isdigit() for char in district):
        return False
    return True
```

```python
for df in [enrol_df, demo_df, biometric_df]:
    df['valid_district'] = df['district_norm'].apply(is_valid_district_name)
```

```python
print("Invalid district rows:")
print("Enrolment:", (~enrol_df['valid_district']).sum())
print("Demographic:", (~demo_df['valid_district']).sum())
print("Biometric:", (~biometric_df['valid_district']).sum())
```

```python
for df in [enrol_df, demo_df, biometric_df]:
    df['district_clean'] = df['district_norm'].where(df['valid_district'], pd.NA)
```

```python
for df in [enrol_df, demo_df, biometric_df]:
    df['analysis_ready'] = (
        df['has_valid_date'] &
        df['valid_state'] &
        df['valid_district']
    )
```

```python
print("Final ANALYSIS-READY unique districts:")
print("Enrolment:", enrol_df.loc[enrol_df['analysis_ready'], 'district_clean'].nunique())
print("Demographic:", demo_df.loc[demo_df['analysis_ready'], 'district_clean'].nunique())
print("Biometric:", biometric_df.loc[biometric_df['analysis_ready'], 'district_clean'].nunique())
```
